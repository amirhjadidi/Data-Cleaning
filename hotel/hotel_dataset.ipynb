```python
import pandas as pd
```

```python
data = pd.read_csv("hotel_bookings.csv", encoding="UTF-8")
print(f"dataset loaded with {data.shape[0]} rows and {data.shape[1]} columns")
```

# 1. rename or drop columns


```python
data.columns.tolist()
```

```python
data.rename(
    {
        "adults": "number_of_adults",
        "children": "number_of_children",
        "babies": "number_of_babies",
    },
    axis=1,
    inplace=True,
)
```

```python
data.columns.tolist()
```

# 2. Handling NaN values


```python
# Check the Percentage of Missing Data in Each Column
data.isna().sum() * 100 / len(data)
```

```python
data.drop("company", axis=1, inplace=True)
```

```python
data["agent"] = data["agent"].fillna(-1)
```

```python
data["country"] = data["country"].fillna("unknown")
```

```python
data.dropna(subset="number_of_children", inplace=True)
```

```python
data.isna().sum() * 100 / len(data)
```

# 3. data type Checks


```python
data.info()
```

```python
data["number_of_children"] = data["number_of_children"].astype("int")
data["agent"] = data["agent"].astype("int")

data = data.astype({"is_canceled": "boolean", "is_repeated_guest": "boolean"})
```
