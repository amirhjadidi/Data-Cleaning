```
import pandas as pd
```

```
data = pd.read_excel('./Customer Call List (1).xlsx')
print(f'dataset loaded with {data.shape[0]} rows and {data.shape[1]} columns')
```

```
data.info()
```

```
data.head()
```

# 1. Drop Duplicates

```
# find duplicates
data[data.duplicated(keep=False)]
```

```
# drop duplicates
data = data.drop_duplicates()
```

```
data[data.duplicated(keep=False)]
```

# 2. Delete Not useful columns

```
data.columns.to_list()
```

```
data.drop('Not_Useful_Column', axis=1, inplace=True)
```

```
data.columns.to_list()
```

# 3. Clean last names

```
data['Last_Name']
```

```
data['Last_Name'] = data['Last_Name'].str.replace(r'^[^a-zA-Z]+|[^a-zA-Z]+$', '', regex=True)
```

# 4. reFormatting phone numbers

```
data['Phone_Number']
```

```
data['Phone_Number'] = data['Phone_Number'].astype('str').str.replace(r'[^0-9]', '', regex=True)
data['Phone_Number']
```

```
data['Phone_Number'] = data['Phone_Number'].replace('nan', pd.NA)
```

```
def format_phone(phone):
    if pd.isna(phone) or phone == '':
        return pd.NA
    if len(phone) == 10:
        return f'{phone[:3]}-{phone[3:6]}-{phone[6:]}'
    return phone


data['Phone_Number'] = data['Phone_Number'].apply(format_phone)
```

```
data['Phone_Number']
```

# 5. reFormatting Address

```
data['Address']
```

```
data['Address'].str.split(',', n=2, expand=True)
```

```
data[['Street_Address', 'State', 'Zip_Code']] = data['Address'].str.split(',', n=2, expand=True)
```

# 6. masking Boolean Columns

```
data[['Paying Customer', 'Do_Not_Contact']]
```

```
mask = {
    'Yes': True,
    'Y': True,
    'No': False,
    'N': False
}
data['Paying Customer'] = data['Paying Customer'].map(mask)
data['Do_Not_Contact'] = data['Do_Not_Contact'].map(mask)
```

```
data[['Paying Customer', 'Do_Not_Contact']]
```

# 7. handling N/A and NAN values

```
data = data.fillna('')
```

```
data
```

# 8. handling if someone doesnt want to call

```
# for x in data.index:
#     if data.loc[x, 'Do_Not_Contact'] == True:
#         data.drop(x, inplace=True)
#     if data.loc[x, 'Do_Not_Contact'] == '':
#         data.drop(x, inplace=True)
#     if data.loc[x, 'Phone_Number'] == '':
#         data.drop(x, inplace=True)

data = data[~((data['Do_Not_Contact'] == True) | (data['Do_Not_Contact'] == '') | (data['Phone_Number'] == ''))]
```

```
data.reset_index(drop=True, inplace=True)
```
